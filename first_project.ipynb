# **Load Data**



```python
import pandas as pd
df=pd.read_csv('https://raw.githubusercontent.com/aneshraj-d96/House-Price-Prediction/refs/heads/main/house_price_prediction.csv')
df = df.dropna()
df
```

# Data Preparation

### Data seperation as x and y


```python
y=df['price']
y
```

```python
x=df.drop(['price', 'address'],axis=1)
x
```

### Data splitting


```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
x_train
```

```python
x_test
```

# Model Building

### Linear Regression

#### Training the model

```python
from sklearn.linear_model import LinearRegression
lr=LinearRegression()
lr.fit(x_train,y_train)
```

### Applying the model to make Prediction

```python
y_lr_train_pred=lr.predict(x_train)
y_lr_test_pred=lr.predict(x_test)
```

```python
print(y_lr_train_pred)
```

```python
print(y_lr_test_pred)
```

## Evaluate Model Performance

```python
from sklearn.metrics import mean_squared_error,r2_score
lr_train_mse=mean_squared_error(y_train,y_lr_train_pred)
lr_train_r2=r2_score(y_train,y_lr_train_pred)

lr_test_mse=mean_squared_error(y_test,y_lr_test_pred)
lr_test_r2=r2_score(y_test,y_lr_test_pred)
```

```python
print(f'LR MSE(Train): {lr_train_mse}')
print(f'LR r2(Train): {lr_train_r2}')
print(f'LR MSE(Test): {lr_test_mse}')
print(f'LR r2(Test): {lr_test_r2}')
```

```python
lr_train_mse
```

```python
lr_train_r2
```

```python
lr_test_mse
```

```python
lr_test_r2
```

```python
lr_results=pd.DataFrame(['Linear Regression',lr_train_mse,lr_train_r2,lr_test_mse,lr_test_r2]).transpose()
lr_results.columns=['Method','Training MSE','Training R2','Testing MSE','Testing R2']
```

```python
lr_results
```

## Random Forest

### Training the Model

```python
from sklearn.ensemble import RandomForestRegressor
rf=RandomForestRegressor(max_depth=2,random_state=42)
rf.fit(x_train,y_train)
```

#### Applying the Model to make prediction

```python
y_rf_train_pred=rf.predict(x_train)
y_rf_test_pred=rf.predict(x_test)
```

### Evaluate model performance

```python
from sklearn.metrics import mean_squared_error,r2_score
rf_train_mse=mean_squared_error(y_train,y_rf_train_pred)
rf_train_r2=r2_score(y_train,y_rf_train_pred)

rf_test_mse=mean_squared_error(y_test,y_rf_test_pred)
rf_test_r2=r2_score(y_test,y_rf_test_pred)
```

```python
rf_results=pd.DataFrame(['Random Forest',rf_train_mse,rf_train_r2,rf_test_mse,rf_test_r2]).transpose()
rf_results.columns=['Method','Training MSE','Training R2','Testing MSE','Testing R2']
```

```python
rf_results
```

# Comparing Models

```python
df_models=pd.concat([lr_results,rf_results],axis=0)
df_models
```

```python
df_models.reset_index(drop=True)
```

# Data regression of prediction results

```python
import matplotlib.pyplot as plt
import numpy as np
plt.figure(figsize=(5,5))
plt.scatter(x=y_train,y=y_lr_train_pred,alpha=0.3)
z=np.polyfit(y_train,y_lr_train_pred,1)
p=np.poly1d(z)
plt.plot()
plt.plot(y_train,p(y_train),"r--")
plt.ylabel('Predicted Price')
plt.xlabel('Actual Price  ')
plt.title('Linear Regression')
```
